# Step 3: Operation

## 8. Create the analysis DataFrame

Select the features needed for the dashboard and use `.copy()` to create a DataFrame named `analysis`. Handle missing values in a way that is appropriate for the question. If values need clearer labels or corrected storage types, make those changes here.



```python
analysis = cars[
    ["Name", "Year", "Miles", "Price"]
].copy()

analysis = analysis[
    analysis["Year"].between(1990, 2026)
].copy()

analysis.shape
```

**How did you handle missing values, and why was that choice appropriate?**
The selected features did not contain any missing values, so no records needed to be removed because of missing data. However, I found invalid model-year values outside a reasonable range. I removed records whose Year was not between 1990 and 2026 rather than attempting to guess the intended year. This left 19,149 valid records for the analysis.

## 9. Create any calculated features

If the dashboard needs a calculated category, date component, percentage, rate, or other calculated feature, create it here. If no calculated feature is needed, leave the code cell blank and explain why.



```python
analysis["Make"] = (
    analysis["Name"]
    .str.split()
    .str[0]
)

analysis.head()
```

**What calculated feature or cleaned label did you create, and how will it support the dashboard?**
I created a Make feature by extracting the first word from each vehicle's Name. For example, "Toyota Camry" becomes "Toyota." This provides a simple categorical feature that can be used as the dashboard's interactive manufacturer filter.

## 10. Prepare the control options

Create the labels and values for the dropdown or radio-button control. The visible labels should be understandable to someone who has not seen the original column names.



```python
top_makes = (
    analysis["Make"]
    .value_counts()
    .head(10)
    .index
    .tolist()
)

make_options = [
    {"label": make, "value": make}
    for make in top_makes
]

make_options
```

# Step 5: Evidence



## 13. Create the first static Plotly figure

Build the first figure before placing it in Dash. Give it a descriptive title and human-readable axis and legend labels.



```python
selected_make = "Toyota"

selected = analysis[
    analysis["Make"] == selected_make
]

fig1 = px.scatter(
    selected,
    x="Miles",
    y="Price",
    hover_name="Name",
    title=f"Mileage vs. Asking Price for {selected_make} Vehicles",
    labels={
        "Miles": "Mileage",
        "Price": "Asking Price ($)"
    }
)

fig1.update_xaxes(range=[0, 125000])
fig1.update_yaxes(range=[0, 105000])

fig1.show()
```

**What evidence should a viewer obtain from the first figure?**
The first figure shows the relationship between vehicle mileage and asking price for the selected manufacturer. It allows the viewer to see whether higher-mileage vehicles generally have lower asking prices and also reveals the amount of variation among individual listings.

## 14. Create the second static Plotly figure

The second figure should add different but related evidence rather than repeat the first figure in another style.



```python
year_summary = (
    selected
    .groupby("Year", as_index=False)["Price"]
    .mean()
)

fig2 = px.line(
    year_summary,
    x="Year",
    y="Price",
    markers=True,
    title=f"Average Asking Price by Model Year for {selected_make} Vehicles",
    labels={
        "Year": "Model Year",
        "Price": "Average Asking Price ($)"
    }
)

fig2.update_xaxes(range=[2009, 2023])
fig2.update_yaxes(range=[0, 40000])

fig2.show()
```

**What does the second figure add to the first figure's evidence?**
The second figure adds information about vehicle age by showing the average asking price for each model year. While the first figure focuses on mileage and individual listings, the second summarizes how asking prices vary across model years for the selected manufacturer.

## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
def create_figures(selected_make):
    filtered = analysis[
        analysis["Make"] == selected_make
    ].copy()

    fig1 = px.scatter(
        filtered,
        x="Miles",
        y="Price",
        hover_name="Name",
        title=f"Mileage vs. Asking Price for {selected_make} Vehicles",
        labels={
            "Miles": "Mileage",
            "Price": "Asking Price ($)"
        }
    )

    fig1.update_xaxes(range=[0, 125000])
    fig1.update_yaxes(range=[0, 105000])

    year_summary = (
        filtered
        .groupby("Year", as_index=False)["Price"]
        .mean()
    )

    fig2 = px.line(
        year_summary,
        x="Year",
        y="Price",
        markers=True,
        title=f"Average Asking Price by Model Year for {selected_make} Vehicles",
        labels={
            "Year": "Model Year",
            "Price": "Average Asking Price ($)"
        }
    )

    fig2.update_xaxes(range=[2009, 2023])
    fig2.update_yaxes(range=[0, 40000])

    return fig1, fig2
```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1("Used Vehicle Price Explorer"),

    html.P(
        "Explore how mileage and model year relate to asking prices "
        "for used vehicles from different manufacturers."
    ),

    html.Label("Select a manufacturer:"),

    dcc.Dropdown(
        id="make-dropdown",
        options=make_options,
        value="Toyota",
        clearable=False
    ),

    dcc.Graph(
        id="mileage-price-graph"
    ),

    dcc.Graph(
        id="year-price-graph"
    ),

    html.H3("Conclusion"),

    html.P(
        "The dashboard shows that asking price generally decreases "
        "as mileage increases, while newer model years generally have "
        "higher average asking prices. The strength and variation of "
        "these patterns differ among manufacturers."
    ),

    html.H3("Limitation"),

    html.P(
        "The listings do not account for every factor that affects "
        "vehicle value, such as trim level, condition, accident history, "
        "optional equipment, or local market conditions. Asking prices "
        "also do not necessarily represent final transaction prices."
    )
])
```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python
@app.callback(
    Output("mileage-price-graph", "figure"),
    Output("year-price-graph", "figure"),
    Input("make-dropdown", "value")
)
def update_dashboard(selected_make):
    return create_figures(selected_make)
```
